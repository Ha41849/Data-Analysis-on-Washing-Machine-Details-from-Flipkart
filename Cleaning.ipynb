```python
import pandas as pd
import numpy as np
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
df = pd.read_csv('scraped_data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df
```

```python
Weight = []
for w in df['Weight_Capacity']:
    r = w.replace(' kg','')
    Weight.append(r)
len(Weight)

df['Weight_Capacity'] = Weight
```

```python
df
```

```python
rate = []
for w in df['Total_no_people_rating']:
    r = w.replace(' Ratings','')
    r = r.replace(',','')
    rate.append(r)
len(rate)

df['Total_no_people_rating'] = rate
```

```python
df['Total_no_people_rating']
```

```python
review = []
for w in df['Total_no_people_reviews']:
    r = w.replace(' Reviews','')
    r = r.replace(',','')
    review.append(r)
len(review)

df['Total_no_people_reviews'] = review
```

```python
df['Total_no_people_reviews']
```

```python
c_warranty = []
for w in df['Comprehensive_Warranty']:
    r = w.replace('Comprehensive','')
    r = r.replace('Years','')
    r = r.replace('years','')
    r = r.replace('Year','')
    r = r.strip()
    c_warranty.append(r)
len(c_warranty)

df['Comprehensive_Warranty'] = c_warranty
```

```python
df['Comprehensive_Warranty']
```

```python
import re
df['Comprehensive_Warranty'] = df['Comprehensive_Warranty'].astype(str).str.extract(r"\('(\d+)'")[0].astype(int)
df['Comprehensive_Warranty']
```

```python
df.head()
```

```python
df['Special_Warranty'].dtype
```

```python
df['Special_Warranty'] = df['Special_Warranty'].str.extract(r'(\d+)')
df['Special_Warranty']
```

```python
df['Price'] = df['Price'].str.replace(',', '', regex=False)
```

```python
df['Price']
```

```python
df['Actual_Price'] = df['Actual_Price'].str.replace(',', '', regex=False)
```

```python
df['Actual_Price']
```

```python
# all unique values first
unique_colors = df['Color_Of_Machine'].dropna().unique()

# Pick 20 random ones (or fewer if fewer exist)
sampled_colors = pd.Series(unique_colors).sample(n=25, replace=True, random_state=42)
print(sampled_colors)
```

```python
df['Color_Of_Machine'] = df['Color_Of_Machine'].str.replace(r',\s*(W|Whi)\b', ', White', regex=True)
df['Color_Of_Machine']
```

## Data Visualization

### Univariate

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.countplot(data=df,x='Machine_Speed',hue='Machine_Speed',palette='dark')
plt.ylabel('Count')
plt.yticks(range(0,10))
plt.grid(axis='y')
plt.title('Washing Maching Machine Speed counts')
plt.show()
```

```python
df.iloc[:,7].plot(kind='hist',bins= 100)
plt.title('Frequency of Machine Speed')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Price'], bins=15, kde=True, color='lightgreen')
plt.title('Price Distribution')
plt.show()
```

```python
df['Price'].value_counts().plot(kind='bar')
plt.xlabel('Price')
plt.yticks(range(0,))
plt.title('price Frequency ')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(df['Price'])
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.countplot(y='Brand_Name', data=df, order=df['Brand_Name'].value_counts().index)
plt.title('Brand Frequency')
plt.show()
```

```python
sns.histplot(data=df,x=df['Actual_Price'],label='Actual Price',kde=True)
sns.histplot(data=df,x=df['Price'],label='Price After Discount',kde=True)
plt.title('Frequency of Actual Price')
plt.xlabel('Actual Price')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
sns.histplot(data=df,x=df['Offer_in_percent'],label='Offer',hue='System_Mode',bins=50)
plt.xticks(rotation=90)
plt.xticks(range(0,80,15))
plt.title('Frequency of Offer')
plt.xlabel('Offer')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot Pie Chart
offer_counts = df['Brand_Name'].value_counts()
max_value = offer_counts.max()
explode = [0.1 if i == max_value else 0 for i in offer_counts]
df['Brand_Name'].value_counts().plot(kind='pie', autopct='%1.1f%%',explode=explode, startangle=60, colormap='viridis')
plt.title('Offer Distribution (%)')
```

```python
sns.kdeplot(df['Price'])
plt.title('Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

### Bivariate

```python
sns.barplot(y=df['Brand_Name'],x=df['Price'])
plt.xticks(rotation=30)
plt.title('Brand vs Price')
plt.ylabel('Brand Name')
plt.xlabel('Price')
plt.show()
```

```python
# Price vs Rating
plt.figure(figsize=(6,4))
sns.scatterplot(x='Rating', y='Price', hue='System_Mode', data=df)
plt.title('Bivariate - Price vs Rating')
plt.show()
```

```python
# Brand vs Rating
plt.figure(figsize=(6,4))
sns.scatterplot(x='Rating', y='Brand_Name', hue='System_Mode', data=df)
plt.title('Bivariate - Brand vs Rating')
plt.xticks(rotation=70)
plt.show()
```

```python
df[df['Machine_Speed']>=500].groupby(['Brand_Name','System_Mode'])['Machine_Speed'].mean().unstack().plot(kind='bar', figsize=(10,6))
plt.xlabel('Brand_Name & System_Mode')
plt.ylabel('Average Machine Speed (RPM)')
plt.title('Average Machine Speed ≥ 500 vs Brand & System Mode')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='System_Mode', y='Price', data=df)
plt.title('Price by System Mode')
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=[np.number])
plt.figure(figsize=(6,4))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

### Multivariate

```python
# Pairplot of selected columns
sns.pairplot(df[['Rating', 'Price', 'Actual_Price', 'Weight_Capacity', 'System_Mode']], hue='System_Mode')
plt.suptitle('Multivariate - Pairplot', y=1.02)
plt.show()
```

```python
# Grouped Bar Plot: Average Price by Brand & System Mode
avg_price = df.groupby(['Brand_Name', 'System_Mode'])['Price'].mean().reset_index()
plt.figure(figsize=(12,5))
sns.barplot(x='Brand_Name', y='Price', hue='System_Mode', data=avg_price)
plt.title('Multivariate - Average Price by Brand & System Mode')
plt.xticks(rotation=90)
plt.show()
```
